# PERBAIKAN CITRA DALAM DOMAIN SPASIAL

### P1. Transformasi Kurva (Citra Negatif)

```python
import numpy as np
import matplotlib.pyplot as plt
from skimage import data, img_as_float

# 1
image_gray_uint8 = data.camera()
image_gray_float = img_as_float(image_gray_uint8)

np.negative_gray = 1.0 -image_gray_float

# 2
image_color_uint8 = data.astronaut()
image_color_float = img_as_float(image_color_uint8)

np.negative_color = 1.0 - image_color_float

# 3
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

#baris 1: gray scale
axes[0, 0].imshow(image_gray_float, cmap='gray')
axes[0, 0].set_title("Grayscale (Original)")
axes[0, 0].axis('off')

axes[0, 1].imshow(np.negative_gray, cmap='gray')
axes[0, 1].set_title("Grayscale (Negative)")
axes[0, 1].axis('off')

# Baris 2: RGB
axes[0, 0].imshow(image_color_float)
axes[0, 0].set_title("Color (Original)")
axes[0, 0].axis('off')

axes[0, 1].imshow(np.negative_color)
axes[0, 1].set_title("Color (Negative)")
axes[0, 1].axis('off')

plt.tight_layout()
plt.show()
```
